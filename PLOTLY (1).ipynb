```python
import numpy as np
import plotly.graph_objects as go

# Generate reproducible random data
np.random.seed(42)
random_x = np.random.randint(1, 101, 100)
random_y = np.random.randint(1, 101, 100)

# Define sizes (replicated/cycled automatically by Plotly for 100 points)
size = [20, 40, 60, 80, 100, 80, 60, 40, 20, 40]

# Build the figure using graph_objects
plot = go.Figure(data=[go.Scatter(
    x = random_x,
    y = random_y,
    mode = 'markers',
    marker = dict(
        size = size, 
        sizemode = 'area',
        sizeref = 2. * max(size) / (40.**2), 
        sizemin = 4
    )
)])

plot.show()
```
