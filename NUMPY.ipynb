```python
a=np.array([[1,2,3,4],[2,4,5,6],[10,20,39,3]])

print("\nPrinting the array:\n")

print(a)

print("\nPrinting the transpose of the array:\n") 
at= a.T

print(at)

print("\nIterating over the transposed array\n")

for x in np.nditer(at): 
    print(x, end=' ')

print("\nSorting the transposed array in C-style:\n")

c = at.copy(order = 'C')

print(c)

print("\nIterating over the C-style array:\n") 
for x in np.nditer(c): 
    print(x,end=' ')

d = at.copy(order = 'F')

print(d)

print("Iterating over the F-style array:\n") 
for x in np.nditer(d):
    print(x,end=' ')
```

```python
arr=np.array([0,10,20,30,40,50,120])
print("printing the values of the angles of sin")
sinval=np.sin(arr*np.pi/180)
print(sinval)
print("printing the inverse of sin")
cosec=np.arcsin(sinval)
print(cosec)
print("printing the values of the degree")
print(np.degrees(cosec))

print("printing the values of the angles of cos")
cosval=np.cos(arr*np.pi/180)
print(cosval)
print("printing the inverse of cos")
sec=np.arccos(cosval)
print(sec)
print("printing the values of the degree")
print(np.degrees(sec))

print("printing the values of the angles of tan")
tanval=np.tan(arr*np.pi/180)
print(tanval)
print("printing the inverse of tan")
cot=np.arctan(sinval)
print(cot)
print("printing the values of the degree")
print(np.degrees(cot))
```
